```python
pages = convert_from_path(r'resources\patient_details\pd_2.pdf', poppler_path=r'C:\poppler-26.02.0\Library\bin')
```

```python
import pytesseract
```

```python
import numpy as np
import cv2
from PIL import Image
# def preprocess_image(img):
#     gray = cv2.cvtColor(np.array(img), cv2.COLOR_BGR2GRAY)
#     resized_img = cv2.resize(gray, None, fx=1.5, fy=1.5, interpolation=cv2.INTER_LINEAR)
#     processed_img = cv2.adaptiveThreshold(resized_img, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 61, 11)
#     return processed_img
def preprocess_image(img):
    # Convert PIL Image to grayscale
    gray = cv2.cvtColor(np.array(img), cv2.COLOR_RGB2GRAY)

    # Remove small noise while preserving text edges
    gray = cv2.medianBlur(gray, 3)

    # Upscale image for better OCR
    gray = cv2.resize(
        gray,
        None,
        fx=2.5,
        fy=2.5,
        interpolation=cv2.INTER_CUBIC
    )

    # Adaptive thresholding
    thresh = cv2.adaptiveThreshold(
        gray,
        255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY,
        61,
        11
    )

    # Remove tiny dots/noise
    kernel = np.ones((2, 2), np.uint8)
    thresh = cv2.morphologyEx(
        thresh,
        cv2.MORPH_OPEN,
        kernel
    )

    return thresh
```

```python
img1 = preprocess_image(pages[0])
img2 = preprocess_image(pages[1])
```

```python
pytesseract.pytesseract.tesseract_cmd=r'C:\Program Files\Tesseract-OCR\tesseract.exe'
text_1 = pytesseract.image_to_string(img1, lang='eng')
text_2 = pytesseract.image_to_string(img2, lang='eng')
print(text_1 + text_2)
```
